                  **VECTORIZATION**

Vectorization is a method in machine learning used to process large sets of data in parallel. This is essential for processing extremely large datasets in algorithm training. Given that traditional Python loops do not implement vectorization, Numpy has to be imported, a Python library written in C. This allows for efficient usage of hardware like GPUs.

```python
import time
import random
```

```python
import numpy
array = numpy.array([3,4,56,6,5])
print(type(array))
```

```python
numberset = numpy.ones(10000000)
print(len(numberset))
randoms = numpy.random.default_rng().integers(1,100,10000000)

print(len(randoms)== len(numberset))
```

```python
#calculating manually


newnums = [1 for _ in range(10000000)]
t1 = time.perf_counter()
for count in range(10000000):
  newnums[count] = randoms[count]* numberset[count]
t2 = time.perf_counter()
slow = t2-t1


#calculating using vectorization
t1 = time.perf_counter()
newnums2= randoms * numberset
t2 = time.perf_counter()
fast = t2-t1
print(slow)
print(fast)
print(all(newnums==newnums2))
print(f"Vectorization is {(slow/fast):.2f} times faster than looping")
```
